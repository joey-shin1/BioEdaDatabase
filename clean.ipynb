```python
import pandas as pd
import numpy as np


def universal_clean(df: pd.DataFrame):
    print('# Total pairs:', len(df))
    print('# Repulsive pairs:', len(df.query('TOTAL > 0')))
    print('# Attractive pairs:', len(df.query('TOTAL < 0')))
    df = df.query("TOTAL < 0")
    print("Dispersion from CLS:", df.index[np.isnan(df['DISP'].values)].tolist())
    df.loc[df.index.tolist(), 'DISP'] = np.where(np.isnan(df['DISP'].values), df['CLS_DISP'], df['DISP'])
    df = df.drop(columns=['PREPARATION', 'ligcoo_x','ligcoo_y','ligcoo_z','protcoo_x','protcoo_y','protcoo_z', 'CLS_DISP','resnr_lig','restype_lig','reschain_lig'])
    return df
```

## HBonds

```python
df = pd.read_csv("hbonds/hbond_EDA.csv", index_col=0)
df = df.query("category != 'Charged-Charged'")
df = universal_clean(df)
df.to_csv("hbonds/hbond_EDA_clean.csv")
```

## Hydrophobic

```python
df = pd.read_csv("hydrophobic/hydrophobic_EDA.csv", index_col=0)
mask = [not index.startswith('1qbv') for index in df.index]
df = df[mask]
df = universal_clean(df)
df.to_csv("hydrophobic/hydrophobic_EDA_clean.csv")
```

## Halogen-bonds

```python
df = pd.read_csv("halogen_bonds/halogen_EDA.csv", index_col=0)
df = universal_clean(df)
df.to_csv("halogen_bonds/halogen_EDA_clean.csv")
```

## Pi-cation

```python
df = pd.read_csv("pi_cation/pi_cation_EDA.csv", index_col=0)
df = df.dropna(subset=['group', 'group1', 'group2'])
df = universal_clean(df)
df.to_csv("pi_cation/pi_cation_EDA_clean.csv")
```

## Pi-stack

```python
df = pd.read_csv("pi_stack/pi_stack_EDA.csv", index_col=0)
df = df.dropna(subset=['prot_hetero', 'lig_hetero' , 'group'])
df = universal_clean(df)
df.to_csv("pi_stack/pi_stack_EDA_clean.csv")
```

## Salt-Bridge

```python
df = pd.read_csv("salt_bridge/salt_bridge_EDA.csv", index_col=0)
df2 = pd.read_csv("hbonds/hbond_EDA.csv", index_col=0).query("category == 'Charged-Charged'")

# merge
for index, row in df2.iterrows():
    for col in df2.columns:
        if col in df.columns:
            df.loc[index, col] = row[col]
    df.loc[index, 'protispos'] = int(row['charge1']) > 0
    df.loc[index, 'note'] = 'from hbond'

df = universal_clean(df)
df.to_csv('salt_bridge/salt_bridge_EDA_clean.csv')
```
